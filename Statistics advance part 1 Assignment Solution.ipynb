# Theoretical Questions:

#### 1. What is a random variable in probability theory?

A random variable is a numerical outcome of a random experiment. It assigns numerical values to events in a sample space.

#### 2. What are the types of random variables?

Discrete Random Variable: Takes a finite or countable number of values (e.g., rolling a die).

Continuous Random Variable: Takes an infinite number of values within a range (e.g., height of a person).

#### 3. What is the difference between discrete and continuous distributions

Discrete distribution: Probability is assigned to specific values (e.g., Binomial, Poisson).

Continuous distribution: Probability is assigned over an interval using a probability density function (PDF) (e.g., Normal, Uniform).



#### 4. What are probability distribution functions (PDF)

A probability distribution function (PDF) describes the likelihood of different outcomes for a random variable.

#### 5. How do cumulative distribution functions (CDF) differ from probability distribution functions (PDF))

PDF provides the probability of an exact value.

CDF gives the cumulative probability up to a certain value.

#### 6. What is a discrete uniform distribution?

A discrete uniform distribution assigns equal probability to all possible outcomes (e.g., rolling a fair die).

#### 9. Write a Python function to calculate Z-scores from a dataset and plot them

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

def calculate_and_plot_zscores(data):
    mean = np.mean(data)
    std_dev = np.std(data, ddof=1)
    z_scores = (data - mean) / std_dev


    plt.figure(figsize=(10, 5))
    sns.scatterplot(x=range(len(data)), y=z_scores, color='b', marker='o', label="Z-Scores")
    plt.axhline(y=0, color='r', linestyle='--', label="Mean (Z=0)")
    plt.axhline(y=1, color='g', linestyle='--', label="1 Std Dev (Z=1)")
    plt.axhline(y=-1, color='g', linestyle='--', label="-1 Std Dev (Z=-1)")
    plt.xlabel("Data Index")
    plt.ylabel("Z-score")
    plt.title("Z-score Plot of Dataset")
    plt.legend()
    plt.show()
    
    return z_scores


data = np.array([10, 12, 15, 18, 20, 22, 24, 30, 35, 40])
z_scores = calculate_and_plot_zscores(data)

print("Z-scores:", z_scores)
```

#### 16. Write a Python function to calculate and plot the standard normal distribution (mean = 0, std = 1)

```python
from scipy.stats import norm

def plot_standard_normal():
    x = np.linspace(-4, 4, 1000)
    y = norm.pdf(x, 0, 1)  

    plt.plot(x, y, label="Standard Normal Distribution")
    plt.xlabel("Z-score")
    plt.ylabel("Probability Density")
    plt.title("Standard Normal Distribution (μ=0, σ=1)")
    plt.legend()
    plt.grid()
    plt.show()

plot_standard_normal()
```

#### 19. Implement hypothesis testing using Z-statistics for a sample dataset

```python
from scipy.stats import norm

sample_mean = 980
population_mean = 1000
std_dev = 50
n = 40

z = (sample_mean - population_mean) / (std_dev / np.sqrt(n))
p_value = norm.cdf(z) * 2  

print(f"Z-score: {z}")
print(f"P-value: {p_value}")


alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis.")
else:
    print("Fail to reject the null hypothesis.")
```

#### 20.  Create a confidence interval for a dataset using Python and interpret the result

```python
import scipy.stats as st

data = np.random.normal(100, 15, 50)
confidence_level = 0.95
mean = np.mean(data)
std_error = st.sem(data) 


conf_interval = st.t.interval(confidence_level, len(data)-1, loc=mean, scale=std_error)
print(f"95% Confidence Interval: {conf_interval}")
```

#### 21.  Generate data from a normal distribution, then calculate and interpret the confidence interval for its mean

```python
data = np.random.normal(100, 15, 50)
mean, std_error = np.mean(data), st.sem(data)

conf_interval = st.t.interval(0.95, len(data)-1, loc=mean, scale=std_error)
print(f"95% Confidence Interval for the Mean: {conf_interval}")
```

#### 27.  Implement Z-score normalization on a dataset and visualize its transformation

```python
def z_score_normalization(data):
    return (data - np.mean(data)) / np.std(data)

data = np.random.normal(100, 20, 100)
normalized_data = z_score_normalization(data)

plt.hist(data, bins=20, alpha=0.5, label="Original Data")
plt.hist(normalized_data, bins=20, alpha=0.5, label="Z-score Normalized Data")
plt.legend()
plt.show()
```
