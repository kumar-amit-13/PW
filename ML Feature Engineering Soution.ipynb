# Feature Engineering

### 1. What is a parameter?

A parameter is a variable in a model that is learned from the data during training. It defines how the model behaves and adjusts to patterns in the dataset.

### 2. What is correlation, What does negative correlation mean?

Correlation is a statistical measure that describes the relationship between two variables. It ranges from -1 to 1, where 1 indicates a strong positive relationship, -1 indicates a strong negative relationship, and 0 means no relationship.

A negative correlation means that as one variable increases, the other decreases. For example, an increase in the price of a product may lead to a decrease in its demand.

### 3. Define Machine Learning. What are the main components in Machine Learning?

Machine Learning is a field of AI that enables computers to learn from data and make predictions. The main components include data, algorithms, models, loss functions, and optimization techniques.

### 4. How does loss value help in determining whether the model is good or not?

The loss value quantifies the error between the predicted and actual outputs. A lower loss value indicates a better model, while a high loss suggests the need for improvements.

### 5. What are continuous and categorical variables?

Continuous variables are numerical and can take infinite values (e.g., height, weight). Categorical variables represent distinct groups or categories (e.g., gender, colors).

### 6. How do we handle categorical variables in Machine Learning? What are the common techniques?

Categorical variables are handled using techniques like One-Hot Encoding, Label Encoding, and Target Encoding to convert them into numerical form for model training.

### 7. What do you mean by training and testing a dataset?

Training a dataset means using data to teach the model, while testing evaluates its performance on unseen data to measure accuracy and generalization.



### 8. What is sklearn.preprocessing?

sklearn.preprocessing is a module in Scikit-learn that provides tools for feature scaling, normalization, encoding categorical variables, and handling missing data.

### 9. What is a Test set?

A test set is a portion of data that is kept separate from training data and is used to evaluate the final model’s performance

### 10. How do we split data for model fitting (training and testing) in Python and How do you approach a Machine Learning problem?

We use train_test_split() from Scikit-learn to split data into training and testing sets, usually in an 80-20 or 70-30 ratio.

The approach includes understanding the problem, collecting and preprocessing data, exploratory data analysis (EDA), feature engineering, model selection, training, evaluation, and optimization.

### 11. Why do we have to perform EDA before fitting a model to the data?

EDA helps understand data distributions, relationships, and missing values, which improves feature selection and model performance.



### 12. What is correlation?

Correlation is a statistical measure that describes the relationship between two variables. It indicates how changes in one variable are associated with changes in another. Correlation values range from -1 to 1

### 13. What does negative correlation mean?

Negative correlation means that as one variable increases, the other variable decreases. It is represented by a correlation coefficient between -1 and 0. A value close to -1 indicates a strong negative relationship, while a value near 0 suggests a weak or no relationship

### 14. How can you find correlation between variables in Python?

In Python, we use df.corr() from Pandas or heatmap() from Seaborn to visualize correlations between numerical variables.

### 15. What is causation? Explain the difference between correlation and causation with an example.

Causation means that one event directly influences another, whereas correlation only shows an association. For example, ice cream sales and drowning deaths are correlated but not causal; both are influenced by hot weather.

### 16. What is an Optimizer? What are different types of optimizers? Explain each with an example.

An optimizer adjusts model parameters to minimize loss. 
Common optimizers include:

SGD (Stochastic Gradient Descent): Updates weights using a random subset of data.

Adam (Adaptive Moment Estimation): Uses momentum and adaptive learning rates for faster convergence.

RMSprop: Suitable for non-stationary objectives by adjusting learning rates dynamically.

### 17. What is sklearn.linear_model?

sklearn.linear_model is a Scikit-learn module for implementing linear models like Linear Regression, Logistic Regression, and Ridge Regression.

### 18. What does model.fit() do? What arguments must be given?

model.fit() trains the model on given data. It requires input features (X) and target values (y).

### 19. What does model.predict() do? What arguments must be given?

model.predict() generates predictions using a trained model. It requires input data (X) for which predictions are needed.

### 20. What are continuous and categorical variables?

Continuous variables are numerical variables that can take an infinite number of values within a given range. Examples include height, weight, temperature, and salary.

Categorical variables represent discrete groups or categories and do not have a numerical meaning. They can be nominal (e.g., gender, color).

### 21. What is feature scaling? How does it help in Machine Learning?

Feature scaling normalizes numerical features to a common scale, improving model performance, especially in distance-based algorithms like KNN and SVM.

### 22. How do we perform scaling in Python?

Scaling is done using StandardScaler for standardization (mean=0, variance=1) and MinMaxScaler for normalization (range 0 to 1) from Scikit-learn.

### 23. What is sklearn.preprocessing?

sklearn.preprocessing is a module in Scikit-Learn that provides tools for preprocessing data before feeding it into a machine learning model.

### 24. How do we split data for model fitting (training and testing) in Python?

In Python, we split data into training and testing sets using train_test_split from sklearn.model_selection. This ensures that the model is trained on one subset and evaluated on another to prevent overfitting.

### 25. Explain data encoding.

Data encoding converts categorical variables into numerical format for machine learning models. Techniques include Label Encoding, One-Hot Encoding, and Binary Encoding.
