#### 1. Explain the different types of data (qualitative and quantitative) and provide examples of each. Discuss nominal, ordinal, interval, and ratio scales.

Data is categorized into two main types: qualitative (categorical) and quantitative (numerical).

#### a) Qualitative Data (Categorical Data):
Qualitative data describes characteristics or qualities that cannot be measured numerically but can be categorized based on attributes or labels.
Example - Customer feedback (satisfied, neutral, dissatisfied), Brand names (Nike, Adidas, Puma)

##### Types of Qualitative Data:
Nominal Scale (No Order, No Meaningful Differences)

Data that consists of categories without any inherent order.
Example: Gender (Male, Female, Non-binary).

Data that has a meaningful order or ranking, but the intervals between values are not uniform.
Example: Survey responses (Poor, Fair, Good, Excellent).


#### b) Quantitative Data (Numerical Data)
Quantitative data represents measurable quantities and can be expressed in numbers. Examples- Number of pets owned by individuals (0, 2, 3, 5)

##### Types of Quantitative Data:
Interval Scale (Ordered, Equal Differences, No True Zero)

Numeric data where the difference between values is meaningful, but there is no true zero point.
Example: Temperature in Celsius or Fahrenheit (20°C, 30°C, 40°C), IQ scores.

Ratio Scale (Ordered, Equal Differences, True Zero Exists)

Similar to interval data but with a meaningful zero, allowing for meaningful ratios.
Example: Height (0 cm means no height), Weight (0 kg means no weight).

### 2. What are the measures of central tendency, and when should you use each? Discuss the mean, median, and mode with examples and situations where each is appropriate.

Measures of central tendency describe the center or typical value of a dataset. The three main measures are mean, median, and mode.
1. Mean (Arithmetic Average): The mean is calculated by adding all values in a dataset and dividing by the total number of values.
Example - Consider the salaries of three employees {40000, 35000, 25000}
mean = (40000 + 35000 + 25000) / 3 = 33333.33
Best use of mean is for Normally distributed data without extreme outliers.
2. Median (Middle Value of Ordered Data): 
The median is the middle value in an ordered dataset. If there is an even number of values, the median is the average of the two middle values.
Example -
Odd-numbered dataset:
Salaries: 40,000, 42,000, 50,000, 55,000, 110,000
Median = 50,000 (middle value)

Even-numbered dataset:
Salaries: 40,000, 42,000, 50,000, 55,000, 80,000, 110,000
Median = (50000 + 45000)/2 = 52500
Best use of median is for Skewed distributions or when there are extreme outliers.

3. Mode (Most Frequently Occurring Value)

The mode is the most frequently occurring value in a dataset. A dataset may have one mode , multiple modes, or no mode if all values occur with equal frequency.

Example:
Test Scores: 85, 90, 92, 85, 95, 88, 85
Mode = 85 (appears most frequently).

Best Use of Mode is for categorical data and identifying the most common value.

### 3. Explain the concept of dispersion. How do variance and standard deviation measure the spread of data?

Dispersion refers to the spread or variability of data in a dataset. It helps measure how much individual values deviate from the central tendency (mean, median, or mode). A high dispersion indicates that data points are spread out, while a low dispersion suggests they are clustered around the central value.

1. Variance: Variance measures the average squared deviation of each data point from the mean. It quantifies how spread out the data points are.

2. Standard Deviation: Standard deviation is the square root of variance and represents the average distance of each data point from the mean. It provides a more intuitive measure of dispersion than variance.


### 4. What is a box plot, and what can it tell you about the distribution of data?

A box plot (box-and-whisker plot also) is a graphical representation of the distribution of a dataset. It summarizes key statistical measures, such as the minimum, first quartile (Q1), median (Q2), third quartile (Q3), and maximum. Box plots help visualize data spread, skewness, and outliers.

Minimum (Q0, Lower Whisker) -> The smallest value in the dataset.

First Quartile (Q1, 25th percentile) -> 25% of the data falls below this value.

Median (Q2, 50th percentile) -> The middle value of the dataset.

Third Quartile (Q3, 75th percentile) -> 75% of the data falls below this value.

Maximum (Q4, Upper Whisker)-> The largest value in the dataset.

### 5. Discuss the role of random sampling in making inferences about populations.

Random sampling is a statistical method used to select a subset of individuals from a larger population in a way that ensures each member has an equal chance of being chosen. Since analyzing an entire population is often impractical, random sampling provides a reliable way to estimate population characteristics without bias.

Types of Random Sampling

1. Simple Random Sampling : Each member of the population has an equal probability of being selected.
Example: Selecting 100 students randomly from a school of 1,000 students using a lottery system.

2. Stratified Random Sampling : The population is divided into subgroups based on characteristics (e.g., age, income, gender), and a random sample is taken from each group.
Example: Surveying 50 students from each grade level (freshman, sophomore, junior, senior) in a university.

3. Systematic Random Sampling : Every k-th individual from a list is selected after choosing a random starting point.
Example: Selecting every 10th person from a list of 1,000 employees

4. Cluster Sampling : The population is divided into clusters (groups), and entire clusters are randomly selected instead of individuals.
Example: Randomly selecting five schools from a district and surveying all students in those schools.

### 6. Explain the concept of skewness and its types. How does skewness affect the interpretation of data?

Skewness is a statistical measure that describes the asymmetry of a dataset’s distribution. It indicates whether data values are more concentrated on one side of the mean.

Symmetrical Distribution (No Skewness) -> The left and right sides of the distribution are mirror images.

Skewed Distribution -> Data is unevenly distributed, causing a shift in the mean, median, and mode.

##### Types of Skewness
1. Positive Skewness (Right-Skewed Distribution) : The right tail of the distribution is longer.
Most data values are concentrated on the left side, with a few extreme high values pulling the mean to the right.

2. Negative Skewness (Left-Skewed Distribution) : The left tail of the distribution is longer.
Most data values are concentrated on the right side, with a few extreme low values pulling the mean to the left.

3. Zero Skewness (Symmetric Distribution) : The distribution is perfectly balanced around the mean.

#### How Skewness Affects Data Interpretation

1. Influence on Measures of Central Tendency

In positively skewed data, the mean is greater than the median, leading to an overestimation of the central value.

2. Impact on Decision-Making

If income data is right-skewed, using the mean salary might misrepresent the financial situation of most people.

3. Effect on Statistical Analysis

Many statistical tests (e.g., t-tests, ANOVA) assume normal distribution (zero skewness).

### 7. What is the interquartile range (IQR), and how is it used to detect outliers?

The Interquartile Range (IQR) is a measure of statistical dispersion that represents the range within which the middle 50% of a dataset falls. It helps identify how spread out the central portion of the data is. An observation is considered an outlier if it falls outside the range of Lower Bound = Q1-1.5*IQR and Upper Bound = Q3+1.5*IQR
Any value below the Lower Bound or above the Upper Bound is an outlier.

#### 9. Explain the Properties of the Normal Distribution and the Empirical Rule (68-95-99.7 Rule).

The normal distribution is a symmetrical, bell-shaped distribution described by its mean (μ) and standard deviation (σ).

1. Symmetry: The mean, median, and mode are the same.
2. Bell-Shaped Curve: The distribution tails approach zero but never touch the x-axis.
3. Empirical Rule (68-95-99.7 Rule):

68% of data falls within 1 standard deviation

95% of data falls within 2 standard deviations

99.7% of data falls within 3 standard deviations

#### 10. Provide a Real-Life Example of a Poisson Process and Calculate the Probability for a Specific Event.

A Poisson distribution models the number of times an event occurs in a fixed interval of time or space.

Example:
A call center receives an average of 5 calls per hour. What is the probability of receiving exactly 3 calls in an hour?

```python
from scipy.stats import poisson
lambda_= 5
k = 3

prob = poisson.pmf(k, lambda_)
print(f"Probability of receiving exactly 3 calls: {prob:.4f}")
```
