### 1. Write a Python program to perform a Z-test for comparing a sample mean to a known population mean and interpret the results

```python
sample = [52, 48, 49, 51, 50, 47, 53, 49, 50, 48]

population_mean = 50 # meu0   
population_std = 2     
alpha = 0.05            

sample_mean = np.mean(sample)
sample_size = len(sample)
standard_error = population_std / np.sqrt(sample_size)
z_score = (sample_mean - population_mean) / standard_error
p_value = 2 * (1 - stats.norm.cdf(abs(z_score))) # Calculating two-tailed p-value

print("Sample mean:", round(sample_mean, 2))
print("Z-score:", round(z_score, 3))
print("P-value:", round(p_value, 4))

if p_value < alpha:
    print("Reject the null hypothesis: the sample mean is significantly different from the population mean.")
else:
    print("Fail to reject the null hypothesis: no significant difference from the population mean.")
```

### 4. Perform a two-tailed Z-test using Python and visualize the decision region on a plot

```python
sample = [51, 49, 52, 50, 48, 53, 47, 50, 49, 51]

population_mean = 50
population_std = 2
alpha = 0.05 

sample_mean = np.mean(sample)
n = len(sample)
standard_error = population_std / np.sqrt(n)

z_score = (sample_mean - population_mean) / standard_error
p_value = 2 * (1 - stats.norm.cdf(abs(z_score)))
z_critical = stats.norm.ppf(1 - alpha / 2)

print("Sample mean:", round(sample_mean, 2))
print("Z-score:", round(z_score, 3))
print("P-value:", round(p_value, 4))

if abs(z_score) > z_critical:
    print("Reject H0: Sample mean is significantly different.")
else:
    print("Fail to reject H0: No significant difference.")

# Ploting standard normal distribution
x = np.linspace(-4, 4, 1000)
y = stats.norm.pdf(x, 0, 1)

plt.figure(figsize=(10, 5))
plt.plot(x, y, label='Standard Normal Distribution', color='blue')

# Shade rejection regions
plt.fill_between(x, y, where=(x <= -z_critical), color='red', alpha=0.5, label='Rejection Region')
plt.fill_between(x, y, where=(x >= z_critical), color='red', alpha=0.5)

# Plot Z-score
plt.axvline(z_score, color='black', linestyle='--', linewidth=2, label=f'Z-score = {z_score:.2f}')

# Labels and legend
plt.title('Two-Tailed Z-Test Visualization')
plt.xlabel('Z-value')
plt.ylabel('Probability Density')
plt.legend()
plt.grid(True)
plt.show()
```

### 5. Create a Python function that calculates and visualizes Type 1 and Type 2 errors during hypothesis testing

```python

def visualize_type1_type2_errors(mu0, mu1, sigma, n, alpha=0.05):
    standurd_error = sigma / np.sqrt(n)
    z_critical = stats.norm.ppf(1 - alpha)
    x_critical = mu0 + z_critical * standurd_error 

   
    beta = stats.norm.cdf((x_critical - mu1) / standurd_error)  # Beta: P(fail to reject H0 | H1 is true)

    x = np.linspace(mu0 - 4 * standurd_error, mu1 + 4 * standurd_error, 1000)
    y0 = stats.norm.pdf(x, mu0, standurd_error)  # Under H0
    y1 = stats.norm.pdf(x, mu1, standurd_error)  # Under H1

    plt.figure(figsize=(10, 6))
    plt.plot(x, y0, label="H₀: μ = {:.2f}".format(mu0), color='blue')
    plt.plot(x, y1, label="H₁: μ = {:.2f}".format(mu1), color='green')

    # Shading Type I error region (alpha)
    x_alpha = np.linspace(x_critical, x[-1], 300)
    plt.fill_between(x_alpha, stats.norm.pdf(x_alpha, mu0, standurd_error), color='red', alpha=0.5, label="Type I error (α)")

    # Shading Type II error region (beta)
    x_beta = np.linspace(x[0], x_critical, 300)
    plt.fill_between(x_beta, stats.norm.pdf(x_beta, mu1, standurd_error), color='orange', alpha=0.5, label="Type II error (β)")

    plt.axvline(x_critical, color='black', linestyle='--', label='Critical value')

    plt.title("Type I and Type II Errors")
    plt.xlabel("Sample Mean")
    plt.ylabel("Probability Density")
    plt.legend()
    plt.grid(True)
    plt.show()

    print(f"Critical value (sample mean): {x_critical:.2f}")
    print(f"Type I error (α): {alpha:.3f}")
    print(f"Type II error (β): {beta:.3f}")

mu0=100
mu1=105
sigma=15
n=30
alpha=0.05

visualize_type1_type2_errors(mu0, mu1, sigma, n, alpha)
```

### 6. Write a Python program to perform an independent T-test and interpret the results

```python

def perform_independent_ttest(alpha=0.05):
    # Generating random data for two groups
    np.random.seed(42)  # for reproducibility
    group1 = np.random.normal(loc=100, scale=15, size=50) 
    group2 = np.random.normal(loc=110, scale=15, size=50) 

    # Performing independent t-test
    t_stat, p_value = stats.ttest_ind(group1, group2)

    print("Group 1 Mean:", np.mean(group1))
    print("Group 2 Mean:", np.mean(group2))
    print("T-statistic:", t_stat)
    print("P-value:", p_value)

    if p_value < alpha:
        print(f"Since p-value < {alpha}, we reject the null hypothesis.")
        print("Conclusion: There is a statistically significant difference between the two groups.")
    else:
        print(f"Since p-value >= {alpha}, we fail to reject the null hypothesis.")
        print("Conclusion: No statistically significant difference between the two groups.")

perform_independent_ttest()
```

### 9. Write a Python function to calculate the confidence interval for a sample mean and explain its significance

```python
def confidence_interval(data, confidence=0.95):
    data = np.array(data)
    n = len(data)
    mean = np.mean(data)
    std_err = stats.sem(data) 
    margin = stats.t.ppf((1 + confidence) / 2, df=n-1) * std_err
    lower = mean - margin
    upper = mean + margin
    return mean, lower, upper


np.random.seed(0)
sample_data = np.random.normal(loc=100, scale=15, size=30)
mean, lower, upper = confidence_interval(sample_data, confidence=0.95)

print(f"Sample Mean: {mean:.2f}")
print(f"95% Confidence Interval: ({lower:.2f}, {upper:.2f})")
```

### 10. Write a Python program to calculate the margin of error for a given confidence level using sample data

```python
def calculate_margin_of_error(data, confidence=0.95):
    data = np.array(data)
    n = len(data)
    std_err = stats.sem(data)
    t_critical = stats.t.ppf((1 + confidence) / 2, df=n-1)
    margin_of_error = t_critical * std_err
    return margin_of_error


np.random.seed(42)
sample_data = np.random.normal(loc=75, scale=10, size=30)
moe = calculate_margin_of_error(sample_data, confidence=0.95)

print(f"Sample Size: {len(sample_data)}")
print(f"Sample Mean: {np.mean(sample_data):.2f}")
print(f"Margin of Error (95% CI): ±{moe:.2f}")
```

### 11. Implement a Bayesian inference method using Bayes' Theorem in Python and explain the process

```python
def bayesian_inference(prior, likelihood, false_positive):
    not_disease = 1 - prior
    evidence = (likelihood * prior) + (false_positive * not_disease)  # Total probability of a positive test
    posterior = (likelihood * prior) / evidence  # Bayes' Theorem PP(Positive∣Disease)
    return posterior


prior = 0.01             
likelihood = 0.99      
false_positive = 0.05   

posterior = bayesian_inference(prior, likelihood, false_positive)
print(f"Probability of having the disease given a positive test: {posterior:.4f}")
```

### 12. Perform a Chi-square test for independence between two categorical variables in Python

```python
def chi_square_independence_test():
    data = {
        'Product A': [30, 20], #Creating a contingency table by gender
        'Product B': [10, 40]
    }
    table = pd.DataFrame(data, index=['Male', 'Female'])
    print("Contingency Table:\n",table)

    chi2, p, dof, expected = chi2_contingency(table)

    print("\nChi-square Test Results:")
    print(f"Chi2 Statistic = {chi2:.4f}")
    print(f"Degrees of Freedom = {dof}")
    print(f"P-value = {p:.4f}")
    print("\nExpected Frequencies:")
    print(pd.DataFrame(expected, index=table.index, columns=table.columns))

    alpha = 0.05
    if p < alpha:
        print("\nConclusion: Reject H0 → There is a significant association between the variables.")
    else:
        print("\nConclusion: Fail to reject H0 → No significant association between the variables.")

chi_square_independence_test()
```

### 13. Write a Python program to calculate the expected frequencies for a Chi-square test based on observed data

```python
def calculate_expected_frequencies(observed):
    # Convert to numpy array
    observed_values = observed.values
    row_totals = observed_values.sum(axis=1, keepdims=True)
    col_totals = observed_values.sum(axis=0, keepdims=True)
    grand_total = observed_values.sum()

    expected = (row_totals @ col_totals) / grand_total
    expected_df = pd.DataFrame(expected, index=observed.index, columns=observed.columns)
    return expected_df



data = {
    'Product A': [30, 20], # Observed data (contingency table)
    'Product B': [10, 40]
}
observed_df = pd.DataFrame(data, index=['Male', 'Female'])

print("Observed Frequencies:")
print(observed_df)

expected_df = calculate_expected_frequencies(observed_df)

print("\nExpected Frequencies:")
print(expected_df)
```

### 14. Perform a goodness-of-fit test using Python to compare the observed data to an expected distribution

```python
def goodness_of_fit_test(observed, expected_probs):
    observed = np.array(observed)
    expected_probs = np.array(expected_probs)

    if not np.isclose(expected_probs.sum(), 1):
        raise ValueError("Expected probabilities must sum to 1.")
    total = observed.sum()
    expected_freq = total * expected_probs

    chi2_stat, p_value = stats.chisquare(f_obs=observed, f_exp=expected_freq)

    return chi2_stat, p_value


observed_counts = [50, 30, 20]                
expected_distribution = [0.4, 0.4, 0.2]       

chi2_stat, p_val = goodness_of_fit_test(observed_counts, expected_distribution)

print(f"Chi-square Statistic: {chi2_stat:.4f}")
print(f"P-value: {p_val:.4f}")

alpha = 0.05
if p_val < alpha:
    print("Reject H₀: The observed data do NOT fit the expected distribution.")
else:
    print("Fail to reject H₀: The observed data fit the expected distribution.")
```

### 16.  Implement an F-test using Python to compare the variances of two random samples

```python
def f_test(sample1, sample2, alpha=0.05):
    var1 = np.var(sample1, ddof=1) 
    var2 = np.var(sample2, ddof=1) 

    if var1 > var2:
        F = var1 / var2
        dfn = len(sample1) - 1  # degrees of freedom numerator
        dfd = len(sample2) - 1  # degrees of freedom denominator
    else:
        F = var2 / var1
        dfn = len(sample2) - 1
        dfd = len(sample1) - 1

    p = 2 * min(stats.f.cdf(F, dfn, dfd), 1 - stats.f.cdf(F, dfn, dfd))

    reject = p < alpha

    return {
        'F_statistic': F,
        'p_value': p,
        'reject_null': reject
    }

np.random.seed(0)
sample_a = np.random.normal(loc=0, scale=1, size=30)
sample_b = np.random.normal(loc=0, scale=1.5, size=35)

result = f_test(sample_a, sample_b)
print(f"F-statistic: {result['F_statistic']:.4f}")
print(f"P-value: {result['p_value']:.4f}")
print("Reject null hypothesis of equal variances:" , result['reject_null'])
```

### 17. Write a Python program to perform an ANOVA test to compare means between multiple groups and interpret the results

```python
def perform_anova(*groups, alpha=0.05):
    F_statistic, p_value = stats.f_oneway(*groups)
    reject_null = p_value < alpha
    interpretation = ("Reject the null hypothesis: at least one group mean is different."
                      if reject_null else
                      "Fail to reject the null hypothesis: no significant difference between group means.")

    return {
        'F_statistic': F_statistic,
        'p_value': p_value,
        'reject_null': reject_null,
        'interpretation': interpretation
    }

np.random.seed(0)
group1 = np.random.normal(loc=20, scale=5, size=30)
group2 = np.random.normal(loc=22, scale=5, size=30)
group3 = np.random.normal(loc=19, scale=5, size=30)

result = perform_anova(group1, group2, group3)

print(f"F-statistic: {result['F_statistic']:.4f}")
print(f"P-value: {result['p_value']:.4f}")
print(result['interpretation'])
```

### 19.  Write a Python function to check the assumptions (normality, independence, and equal variance) for ANOVA

```python
def check_anova_assumptions(*groups, alpha=0.05):
    results = {}
    # 1) Normality test (Shapiro-Wilk) for each group
    normality = {}
    for i, group in enumerate(groups, 1):
        stat, p = stats.shapiro(group)
        normality[f'Group_{i}'] = {
            'W_statistic': stat,
            'p_value': p,
            'normal': p > alpha
        }

    # 2) Equal variances (Levene's test)
    stat, p = stats.levene(*groups)
    equal_variance = {
        'Levene_statistic': stat,
        'p_value': p,
        'equal_variance': p > alpha
    }

    # 3) Independence — assumed based on study design
    independence = "Assumed based on study design; cannot be statistically tested here."

    results['normality'] = normality
    results['equal_variance'] = equal_variance
    results['independence'] = independence

    return results


np.random.seed(42)
group1 = np.random.normal(20, 4, 30)
group2 = np.random.normal(22, 5, 30)
group3 = np.random.normal(19, 4, 30)

assumption_results = check_anova_assumptions(group1, group2, group3)

print("Normality Test Results (Shapiro-Wilk):")
for group, res in assumption_results['normality'].items():
    print(f" {group}: W={res['W_statistic']:.4f}, p={res['p_value']:.4f}, normal={res['normal']}")

print("\nEqual Variance Test (Levene's):")
print(f" Statistic={assumption_results['equal_variance']['Levene_statistic']:.4f}, "
      f"p={assumption_results['equal_variance']['p_value']:.4f}, "
      f"equal_variance={assumption_results['equal_variance']['equal_variance']}")

print("\nIndependence Assumption:")
print(assumption_results['independence'])
```

### 20. Perform a two-way ANOVA test using Python to study the interaction between two factors and visualize the results

### 23. Simulate random data from a normal distribution, then perform hypothesis testing to evaluate the means

```python
def simulate_and_test(sample_size=30, population_mean=50, sample_mean_shift=0, alpha=0.05):
    np.random.seed(42)
    sample = np.random.normal(loc=population_mean + sample_mean_shift, scale=10, size=sample_size)

    # Performing one-sample t-test
    t_stat, p_value = stats.ttest_1samp(sample, population_mean)

    reject_null = p_value < alpha
    interpretation = (
        "Reject the null hypothesis: sample mean is significantly different from population mean."
        if reject_null else
        "Fail to reject the null hypothesis: no significant difference from population mean."
    )

    return {
        'sample_mean': np.mean(sample),
        't_statistic': t_stat,
        'p_value': p_value,
        'reject_null': reject_null,
        'interpretation': interpretation
    }


result = simulate_and_test(sample_size=40, population_mean=50, sample_mean_shift=3)

print(f"Sample Mean: {result['sample_mean']:.2f}")
print(f"T-statistic: {result['t_statistic']:.4f}")
print(f"P-value: {result['p_value']:.4f}")
print(result['interpretation'])
```

### 24. Perform a hypothesis test for population variance using a Chi-square distribution and interpret the results

```python
def chi_square_variance_test(sample, sigma_0_squared, alpha=0.05):
    n = len(sample)
    sample_var = np.var(sample, ddof=1) 
    chi2_stat = (n - 1) * sample_var / sigma_0_squared

    chi2_critical_low = stats.chi2.ppf(alpha / 2, df=n - 1)
    chi2_critical_high = stats.chi2.ppf(1 - alpha / 2, df=n - 1)

    reject_null = chi2_stat < chi2_critical_low or chi2_stat > chi2_critical_high
    interpretation = (
        "Reject the null hypothesis: sample variance differs significantly from hypothesized variance."
        if reject_null else
        "Fail to reject the null hypothesis: no significant difference in variance."
    )

    return {
        'sample_variance': sample_var,
        'chi2_statistic': chi2_stat,
        'chi2_critical_low': chi2_critical_low,
        'chi2_critical_high': chi2_critical_high,
        'reject_null': reject_null,
        'interpretation': interpretation
    }

np.random.seed(0)
sample_data = np.random.normal(loc=50, scale=5, size=30)  # true variance = 25
sigma_0_squared = 36  

result = chi_square_variance_test(sample_data, sigma_0_squared)

print(f"Sample Variance: {result['sample_variance']:.4f}")
print(f"Chi-square Statistic: {result['chi2_statistic']:.4f}")
print(f"Critical Values: Lower={result['chi2_critical_low']:.4f}, Upper={result['chi2_critical_high']:.4f}")
print(result['interpretation'])
```

### 25. Write a Python script to perform a Z-test for comparing proportions between two datasets or groups

```python
def z_test_proportions(x1, n1, x2, n2, alpha=0.05):
    p1 = x1 / n1
    p2 = x2 / n2
    p_pool = (x1 + x2) / (n1 + n2)
    se = np.sqrt(p_pool * (1 - p_pool) * (1/n1 + 1/n2)) # se- standurd error
    z = (p1 - p2) / se     # Z statistic
    p_value = 2 * (1 - stats.norm.cdf(abs(z)))

    reject_null = p_value < alpha
    interpretation = (
        "Reject the null hypothesis: proportions differ significantly."
        if reject_null else
        "Fail to reject the null hypothesis: no significant difference in proportions."
    )

    return {
        'z_score': z,
        'p_value': p_value,
        'reject_null': reject_null,
        'interpretation': interpretation
    }

result = z_test_proportions(x1=40, n1=100, x2=60, n2=120)

print(f"Z-score: {result['z_score']:.4f}")
print(f"P-value: {result['p_value']:.4f}")
print(result['interpretation'])
```

### 26. Implement an F-test for comparing the variances of two datasets, then interpret and visualize the results

```python
def f_test_variance(sample1, sample2, alpha=0.05, visualize=True):
    n1, n2 = len(sample1), len(sample2)


    var1, var2 = np.var(sample1, ddof=1), np.var(sample2, ddof=1)    # Sample variances

    # Arrange so that F ≥ 1
    if var1 > var2:
        F = var1 / var2
        dfn, dfd = n1 - 1, n2 - 1
    else:
        F = var2 / var1
        dfn, dfd = n2 - 1, n1 - 1

    # Two-tailed test critical values
    F_crit_low = stats.f.ppf(alpha / 2, dfn, dfd)
    F_crit_high = stats.f.ppf(1 - alpha / 2, dfn, dfd)
    p_value = 2 * min(stats.f.cdf(F, dfn, dfd), 1 - stats.f.cdf(F, dfn, dfd))

    reject_null = F < F_crit_low or F > F_crit_high
    interpretation = (
        "Reject the null hypothesis: variances are significantly different."
        if reject_null else
        "Fail to reject the null hypothesis: no significant difference in variances."
    )

    if visualize:
        x = np.linspace(0.01, 5, 500)
        y = stats.f.pdf(x, dfn, dfd)
        plt.figure(figsize=(8, 5))
        plt.plot(x, y, 'b-', lw=2, label='F-distribution')
        plt.axvline(F, color='r', linestyle='--', label=f'F statistic = {F:.2f}')
        plt.axvline(F_crit_low, color='g', linestyle='--', label=f'Lower critical = {F_crit_low:.2f}')
        plt.axvline(F_crit_high, color='g', linestyle='--', label=f'Upper critical = {F_crit_high:.2f}')
        plt.title('F-test for Equality of Variances')
        plt.xlabel('F value')
        plt.ylabel('Density')
        plt.legend()
        plt.grid(True)
        plt.show()

    return {
        'F_statistic': F,
        'degrees_of_freedom': (dfn, dfd),
        'p_value': p_value,
        'critical_values': (F_crit_low, F_crit_high),
        'reject_null': reject_null,
        'interpretation': interpretation
    }


np.random.seed(1)
group1 = np.random.normal(10, 2, 30)
group2 = np.random.normal(10, 3, 30)

result = f_test_variance(group1, group2)

print(f"F-statistic: {result['F_statistic']:.4f}")
print(f"Degrees of freedom: {result['degrees_of_freedom']}")
print(f"Critical values: {result['critical_values']}")
print(f"P-value: {result['p_value']:.4f}")
print(result['interpretation'])
```

### 26. Perform a Chi-square test for goodness of fit with simulated data and analyze the results

```python
def chi_square_goodness_of_fit(observed, expected, alpha=0.05):
    chi2_stat, p_value = stats.chisquare(f_obs=observed, f_exp=expected)

    reject_null = p_value < alpha
    interpretation = (
        "Reject the null hypothesis: observed data does not fit the expected distribution."
        if reject_null else
        "Fail to reject the null hypothesis: observed data fits the expected distribution."
    )

    return {
        'chi2_statistic': chi2_stat,
        'p_value': p_value,
        'reject_null': reject_null,
        'interpretation': interpretation
    }

# Simulating observed frequencies
np.random.seed(42)
categories = ['A', 'B', 'C', 'D']
observed = np.random.multinomial(n=100, pvals=[0.25, 0.25, 0.25, 0.25])  # equal probs

# Defining expected frequencies (uniform)
expected = np.array([25, 25, 25, 25])  # expected under uniform distribution

result = chi_square_goodness_of_fit(observed, expected)

print("Observed:", observed)
print("Expected:", expected)
print(f"Chi-square Statistic: {result['chi2_statistic']:.4f}")
print(f"P-value: {result['p_value']:.4f}")
print(result['interpretation'])

plt.figure(figsize=(8, 5))
bar_width = 0.35
x = np.arange(len(categories))
plt.bar(x, observed, width=bar_width, label='Observed', color='skyblue')
plt.bar(x + bar_width, expected, width=bar_width, label='Expected', color='orange')
plt.xticks(x + bar_width / 2, categories)
plt.title("Chi-square Goodness-of-Fit Test")
plt.xlabel("Category")
plt.ylabel("Frequency")
plt.legend()
plt.tight_layout()
plt.grid(True, axis='y', linestyle='--', alpha=0.7)
plt.show()
```
